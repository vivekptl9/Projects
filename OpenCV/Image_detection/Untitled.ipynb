```python
image = cv2.imread("Xeg12GyGR5KGwsWtQVkO")
```

```python
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
pyplot.imshow(image)
```

```python
image = cv2.resize(image, (0, 0), fx=0.75, fy=0.75)
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
gray_image = cv2.cvtColor(gray_image, cv2.COLOR_BGR2RGB)
```

```python

pyplot.imshow(gray_image)
```

```python
gray_image = cv2.cvtColor(gray_image, cv2.COLOR_BGR2RGB)
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

MAXIMUM_CORNERS = 100

QUALITY_LEVEL = 0.01

MINIMUM_DISTANCE = 10

corners = cv2.goodFeaturesToTrack(gray_image, 
                                  MAXIMUM_CORNERS, 
                                  QUALITY_LEVEL,
                                  MINIMUM_DISTANCE)
```

```python
#corners
```

```python
import numpy
corners = numpy.int64(corners)
RADIUS = 5

COLOR = (255, 255, 32)

for corner in corners:

  x, y = corner.ravel()

  cv2.circle(image, (x, y), RADIUS, COLOR, -1)
```
